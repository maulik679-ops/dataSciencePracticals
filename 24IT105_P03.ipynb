```python
import pandas as pd
df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/Pract_3/city_day.csv")
df.head()
```

```python
print("Rows, Columns:", df.shape)
df.info()
# Descriptive stats
df.describe()
```

```python
# % missing per column
missing = df.isnull().mean() * 100
print(missing)

# Simple imputation (fill numeric nulls with median)
numeric_cols = df.select_dtypes(include=['float64','int64']).columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].median())

# Fill categorical nulls
df["AQI_Bucket"] = df["AQI_Bucket"].fillna("Unknown")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

corr = df[numeric_cols].corr()

plt.figure(figsize=(12, 10))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation Heatmap (Pollutants)")
plt.show()
```

```python
df["Date"] = pd.to_datetime(df["Date"])
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month

pm25_yearly = df.groupby("Year")["PM2.5"].mean().reset_index()

sns.lineplot(data=pm25_yearly, x="Year", y="PM2.5")
plt.title("Annual Average PM2.5 Trend")
plt.show()
```

```python
plt.figure(figsize=(15,6))
sns.boxplot(x="City", y="PM10", data=df)
plt.xticks(rotation=90)
plt.title("PM10 Distribution Across Cities")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["AQI"], bins=30, kde=True)
plt.title("Distribution of AQI Values")
plt.show()
```
